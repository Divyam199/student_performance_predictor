### 9️⃣ Add a Grade Column ➕
#### Q. Create grades based on percentage  ❓

```python
def get_grade(pct):
    if pct >= 90:
        return "A"
    elif pct >= 75:
        return "B"
    elif pct >= 60:
        return "C"
    elif pct >= 40:
        return "D"
    else:
        return "F"

df["Grade"] = df["Percentage"].apply(get_grade)
df
```

### 📊 Student Performance Dashboard

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(7,5))

# Average marks by Subject

axes[0,0].bar(["Maths","Science","English"],
               df[["Maths","Science","English"]].mean(),
               color=["#2C6E98","#27AE60","#F39C12"])
axes[0,0].set_title("Average Marks by Subject")
axes[0,0].set_xlabel("Subjects")
axes[0,0].set_ylabel("Average Marks")

# Pass vs Fail

pf=df["Status"].value_counts()
axes[0,1].pie(pf,labels=pf.index,autopct="%1.0f%%",colors=["#27AE60","#C0392B"])
axes[0,1].set_title("Pass vs Fail")

# Students per Grade

gc = df["Grade"].value_counts().sort_index()
axes[1,0].bar(gc.index,gc.values,color="#7B4C95")
axes[1,0].set_title("Students per Grade")
axes[1,0].set_xlabel("Grades")
axes[1,0].set_ylabel("No. of Students")

# Attendance vs Percentage

axes[1,1].scatter(df["Attendance"],df["Percentage"],color="#F39C12")
axes[1,1].set_title("Attendance vs Percentage")
axes[1,1].set_xlabel("Attendance")
axes[1,1].set_ylabel("Percentage")

plt.suptitle("Stuent Performance Dashboard", fontsize=16)
plt.tight_layout()
#plt.savefig("Student Performance Dashboard.png", dpi=150, bbox_inches="tight") #Uncomment it to save the dashboard chart
plt.show()
```

### 🤖 Using Machine Learning to build Student Pass Predictor

```python
# Importing the Librsries
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
# import joblib # Uncomment if you want to save the model

# Preparing and splitting the data
X = df[["Attendance","Maths","Science","English"]]  #Features
y = df["Status"] #Target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.34, random_state=42, stratify=y)

# Training Model
model = LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)

#Evaluating
pred = model.predict(X_test)
print("Accuracy: ",round(accuracy_score(y_test, pred), 2))
print("Predicted : ", list(pred))
print("Actual : ", list(y_test.values))
print(classification_report(y_test, pred))

# Predicting a new student
new = pd.DataFrame([{"Attendance" : 60, "Maths" : 45, "Science" : 50, "English" : 40}]) #New entry
print("Prediction : ", model.predict(new)[0])

# Saving Model (Uncomment the code below to save)
# joblib.dump(model, "pass_model.pkl")
# print("Model Saved")
```
